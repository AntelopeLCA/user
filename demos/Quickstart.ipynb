## FHWA Resource stored locally

Download `Federal_Highway_Administration-mtu_pavement.zip` from the Federal Commons and save to your computer.

In order to build the study quickly, we will tell the linker to treat ambiguous flows as cut-offs. We can later check for cut-offs and add them to a model.

```python
FHWA = '/data/LCI/FedCommons/Federal_Highway_Administration-mtu_pavement.zip'

# set static=True to load the entire dataset
cat.new_resource('fhwa', FHWA, 'OpenLcaJsonLdArchive', interfaces=('basic', 'exchange', 'index', 'quantity'), static=True)
cat.background_for_origin('fhwa')  # takes about 30 seconds to load the archive

# set multi_term='cutoff' to cut-off ambiguous flows
cat.query('fhwa').check_bg(multi_term='cutoff')  # takes about 10 seconds to create the ordered background

# see our progress
cat.show_interfaces()
```

```python
q = cat.query('fhwa')
q.count('process')
```

## Get LCIA over the air


```python
from antelope import enum
# a visual enumerator, for interactive use 
```

```python
cat.blackbook_guest('https://sc.vault.lc')
cat.get_blackbook_resources('lcia.openlca.2.7.5')
from antelope import enum  # this is a (perhaps poorly-named) utility for enumerating outputs
ts = enum(cat.query('lcia.openlca.2.7.5').lcia(method='TRACI'))
```

```python
ts[0].show()
```

```python
tt = enum(cat.query(ts[0].origin).get(k) for k in ts[0]['ImpactCategories'])
```

### Run LCIA on a process


```python
p = next(cat.query('fhwa').processes(name='asphalt mixture - framework'))
p.bg_lcia(tt[8]).show_details()
```
